### **1. Introdução**
-------------------------------------------------------------------------------------------------------------------------------------------------


No âmbito da Unidade Curricular de Algoritmos e Técnicas de Programação, foi solicitado a implementação de uma aplicação em Python, com o objetivo de gerir uma biblioteca científica que permitisse criar, atualizar e analisar publicações científicas, entre outras funcionalidades. 

A aplicação criada oferece diversas possibilidades aos utlizadores, incluindo o carregamento de uma base de dados, a inserção de novas publicações científicas, consulta das publicações consoante um determinado parâmetro, a remoção de publicações, a atualização de informações das que já constam na base de dados, a listagem das publicações com base em parâmetros específicos, seja o nome do autor da mesma ou através das palavras-chaves que as identificam, a geração de distribuições, a possibilidade de adicionar as publicações aos *Favoritos* e ainda a possibilidade de guardar a base de dados após realizar as alterações.
Para além disso, a nossa Bibliteca Científica dispõe também de um botão *Help* que especifica as funcionalidades de todos os comandos.

Para este projeto, os nossos objetivos consistiam em realizar todas as operações propostas e, para além disso, analisar os múltiplos casos que podem tornar as opções executadas pelo utilizador inválidas. Também era pretendido desenvolver uma interface gráfica, com auxílio do PySimpleGUI, de fácil compreensão e utilização.

#### 2.1 Descrição do Problema e Especificação de Requisitos

Para a realização deste projeto decidimos manter o modelo da base de dados em JSON disponibilizada pelo professor uma vez que se trata de um formato text-based ideal para representar dados estruturados.

Deste modo, a base de dados de publicações científicas consiste numa lista de objetos, os quais representam cada um dos artigos. 
Cada artigo consiste num dicionário cujos conjuntos chave-valor são:

- ***"abstract":***
(string) Representa uma síntese dos objetivos, métodos, resultados e conclusões do artigo científico;

- ***"keywords":***
 (string) Representa os termos-chave dos principais tópicos e conceitos do artigo científico;

- ***"authors":***
 (lista de autores) Cada autor do artigo é representado num dicionário;

- ***"name":`***
 (string) Nome de um autor do artigo;

- ***"affiliation":*** 
(string) Representa a instituição à qual o respetivo autor está vinculado;

- ***"doi":***
 (string) Representa um identificador digital único que facilita o acesso a cada artigo;

- ***"pdf":***
 (string) Representa o formato digital do artigo;

- ***"publish_date":***
 (string) Data da publicação do artigo;

- ***"title":***
 (string) Nome do artigo;

- ***"url":***
 (string) Representa o endereço web que permite acessar diretamente o artigo.

#### 2.2 Funcionalidades Implementadas no Sistema

**Carregar BD**

- Conforme indicado no enunciado, é necessário implementar a capacidade de carregar a base de dados na aplicação. Este carregamento consiste na leitura de um ficheiro que deverá ter a estrutura referida anteriormente e cuja a extensão do mesmo seja JSON. Caso não seja inserida, nenhuma das restantes opções são funcionais.

**Inserir Artigo**

- Com o propósito de garantir que a biblioteca científica exiba plenamente a sua capacidade de gerir e manipular publicações foi elaborada a funcionalidade "Inserir Artigo",
uma função destinada à incorporação de novas publicações à base de dados. O registo de um novo artigo na base de dados é feito através da introdução dos parâmetros obrigatórios correspondentes às chaves do dicionário, tal como explicitado anteriormente. Todos aqueles que não forem inseridos, tomam o valor "Desconhecido2.

**Consultar Artigo**

- Esta funcionalidade permite ao utilizador realizar consultas na base de dados de artigos científicos através de diferentes critérios: *Autor*, *Título*, *Ano*, *Keyword* e *Afiliação do Autor*. A pesquisa é feita de forma flexível e os resultados são apresentados na interface, possibilitando uma visualização detalhada das publicações que atendem aos parâmetros selecionados.

**Eliminar Artigo**

- De forma a que seja permitido ao utilizador remover um artigo da base de dados, adicionamos à aplicação a função "Eliminar Artigo" através do parametro "doi".

**Atualizar Artigo**

- Com o objetivo de permitir que o utilizador consiga proceder à alteração de determinados parâmetros de um certo artigo, nomeadamente da data de publicação, do resumo, palavras-chave, autores e afiliações, implementamos uma funcionalidade denominada "Atualizar Artigo".

**Listagem**

- O programa oferece ainda a funcionalidade de listar artigos com base em dois parâmetros: Autor ou Palavra-Chave. Para cada um desses parâmetros, é possível ordenar os resultados de diferentes maneiras:

    * Ordem alfabética: Para o parâmetro "Autor", os artigos são ordenados pela chave "name", podendo ser de forma crescente (A-Z) ou decrescente (Z-A). O mesmo se aplica ao parâmetro "Palavra-Chave", onde os artigos são ordenados conforme o valor da palavra-chave, também podendo ser de A-Z ou Z-A.

    * Frequência de artigos publicados: Esta opção está disponível apenas para o parâmetro "Autor", permitindo ordenar os artigos com base no número de publicações de cada autor, em ordem crescente ou decrescente.

    * Ocorrências de palavras-chave: Esta opção é específica para o parâmetro "Palavra-Chave". Ela permite ordenar os artigos de acordo com o número de vezes que a palavra-chave aparece nos diferentes artigos da base de dados.

**Distribuição**

- A aplicação permite produzir indicadores estatísticos relativamente aos filmes armazenados, nomeadamente:

    - Distribuição de publicações por ano;

    - Distribuição de publicações por mês de um determinado ano;

    - Número de publicações por autor (top 20 autores);

    - Distribuição de publicações de um autor por anos;

    - Distribuição de palavras-chave pela sua frequência (top 20 palavras-chave);

    - Distribuição de palavras-chave mais frequente por ano.

- Estas distribuições podem ser consultadas sobre o formato gráfico. Para tal foi usada a biblioteca matplotlib que permitiu a representação destes dados em gráficos de barras.

**Guardar Artigos**

- A aplicação possibilita ao utilizador guardar a base de dados num ficheiro.

- Considerando que o formato JSON é o mais adequado para o modelo de dados da aplicação, a funcionalidade de guardar a biblioteca científica atualiza o ficheiro original com o estado atual dos dados.

**Artigos Favoritos**

- A funcionalidade de Favoritos foi implementada para permitir ao utilizador guardar os artigos que considera mais relevantes, tornando-os facilmente acessíveis a qualquer momento. Após realizar uma consulta através da funcionalidade "Consultar Artigo", o utilizador pode adicionar artigos aos seus *favoritos* clicando no botão "Adicionar aos Favoritos".

- Os artigos selecionados são armazenados num ficheiro JSON e podem ser carregados sempre que necessário, oferecendo uma experiência personalizada ao utilizador. Os artigos são mantidos de forma persistente, mesmo após a aplicação ser fechada, garantindo que os favoritos fiquem disponíveis durante as próximas sessões.

- As funções que sustentam esta funcionalidade incluem:

    - Carregar Favoritos: Carrega a lista de artigos favoritos armazenada num ficheiro JSON.

    - Adicionar aos Favoritos: Permite ao utilizador adicionar um artigo à sua lista de favoritos, se ainda não estiver presente.

    - Guardar Favoritos: Assegura que a lista de favoritos seja gravada corretamente no ficheiro JSON após adições ou modificações.

    - Mostrar Favoritos: Permite visualizar o ficheiro com os artigos guardados.

    - Eliminar Favoritos: Que remove os artigos favoritos que ja foram consultados, permitindo adicionar novos.

- Esta funcionalidade proporciona um maior grau de personalização e comodidade, permitindo que os utilizadores guardem artigos importantes para futuras consultas.

**Exportar Dados**

- Tal como a função anterior, também esta funcionalidade está relacionada à "Consultar Artigo". Após proceder corretamente à procura de um artigo consoante o parâmetro pretendido, o utilizador poderá exportar os seus dados para um ficheiro JSON. A aplicação permite-lhe ainda atribuir o nome que pretender a este ficheiro.

**Importar Dados**

- Para além disso, o utilizador poderá ainda importar dados externos à base de dados que carregou inicialmente, caso os deseje incluir no programa para proceder, em conjunto com a base de dados, às alterações e manipulações oferecidas pelas funcionalidades da aplicação.

**Help**

- Com a finalidade de fornecer ao utilizador as indicações para o funcionamento correto de cada comando da nossa aplicação, desenvolvemos a função "Help".
Deste modo, permitimos aprimorar a eficácia da utilização da mesma.

**Sair**

- Assim que o utilizador entender recorrer a esta funcionalidade encerra a aplicação., questionando primeiro se o utilizador que guardar todas as alterações feitas.
